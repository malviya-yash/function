7.  What are the advantages of using generators over regular functions?
    - Generators offer several advantages over regular functions, especially when dealing with large datasets or continuous data streams:

    1. Memory Efficiency

      . Generators yield one value at a time, so they don't store the entire result in memory.

     . Ideal for large data (e.g., reading a big file line by line).

   2. Lazy Evaluation
     
     . Values are produced only when needed, which improves performance.

     . Example: You don’t generate the next item until next() is called.

  3. Inflnite Sequences

     . Generators can model infinite sequences (like counting numbers) without running out of memory.

  4. Clenaner code with less overhead

     . Generator syntax with yield is often simpler and cleaner than manually managing state with classes and iterators.

  5. Better perfomance

     . Since they don’t compute all values at once, they are often faster and more efficient for large or long-running tasks.

```python
#6.Explain the concept of generators in Python and how they are defined.
#Define the Generator:
def count_up_to(n):
    count = 1
    while count <= n:
        yield count  # yields one value at a time
        count += 1
```

```python
#calling the function
counter = count_up_to(3)
print(next(counter))  # 1
print(next(counter))  # 2
print(next(counter))  # 3
```

```python
#4. Write a Python function that checks if a given number is prime or not from 1 to 200.
def is_prime(n):
  if n <= 1:
    return False # number less or equal to 1 are not prime
  for i in range(2, int(n ** 0.5)+1):
    if n % i == 0:
      return False
  return True

prime_numbers = [n for n in range(1, 201)if is_prime(n)]

print("Prime numbers from 1 to 200:", prime_numbers)
```

```python
#5. Create an iterator class in Python that generates the Fibonacci sequence up to a specified number of terms.
class FibonacciIterator:
    def __init__(self, terms):
        self.terms = terms  # Number of terms in the Fibonacci sequence
        self.a, self.b = 0, 1  # Initial two Fibonacci numbers
        self.count = 0  # Counter to track the number of terms generated

    def __iter__(self):
        return self  # Returns the iterator object itself

    def __next__(self):
        if self.count < self.terms:
            # Generate the next Fibonacci number
            fib_number = self.a
            self.a, self.b = self.b, self.a + self.b  # Update the Fibonacci numbers
            self.count += 1
            return fib_number
        else:
            raise StopIteration  # Stop iteration once the specified number of terms is reached

# Example usage:
fibonacci_sequence = FibonacciIterator(10)  # Generate first 10 Fibonacci numbers
for num in fibonacci_sequence:
    print(num)
```

```python
#7. Implement a generator function that reads a file line by line and yields each line as a string.
def read_file_lines(file_path):
    """
    Reads a file line by line and yields each line.

    Args:
        file_path (str): The path to the file.

    Yields:
        str: Each line of the file.
    """
    try:
        with open(file_path, 'r') as file:
            for line in file:
                yield line
    except FileNotFoundError:
        yield f"File not found: {file_path}"
    except Exception as e:
        yield f"An error occurred: {e}"

# Example usage
if __name__ == '__main__':
    # Create a dummy text file for testing
    with open('example.txt', 'w') as f:
        f.write("This is line 1.\n")
        f.write("This is line 2.\n")
        f.write("This is line 3.\n")

    # Use the generator to read the file line by line
    for line in read_file_lines('example.txt'):
        print(line.strip())  # Print each line with leading/trailing whitespace removed

    #Example of handling a non-existent file
    for line in read_file_lines('non_existent_file.txt'):
        print(line)
```
